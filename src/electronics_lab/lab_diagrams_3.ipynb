### Διαγράμματα 3ης Εργαστηριακής Αναφοράς 

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
V_in = np.array([13, 25, 30, 40, 50, 60, -13, -25, -30, -40, -50])
V_out = np.array([1.33, 2.59, 3.10, 4.08, 4.12, 4.12, -1.33, -2.52, -2.82, -2.82, -2.82])

df = pd.DataFrame(
    np.concat([V_in.reshape(-1,1), V_out.reshape(-1, 1)], axis=1),
    columns=["V_in (mV)", "V_out (V)"],
)

df
```

```python
sorted_df = df.sort_values("V_in (mV)")
sorted_df
```

```python
plt.plot(sorted_df["V_in (mV)"], sorted_df["V_out (V)"], "-o")
plt.title("Χαρακτηριστική V-I", fontdict={"size": 20})
plt.xlabel(r"$V_{in} \: (mV)$", fontdict={"size": 14})
plt.ylabel(r"$V_{out} \: (V)$", fontdict={"size": 14})
plt.grid()
plt.tight_layout()
plt.show()
```

```python
V_in_linear = np.array([-25, -13, 13, 25, 30, 40]) / 1000
V_out_linear = np.array([-2.52, -1.33, 1.33, 2.59, 3.10, 4.08])

slope, intercept = np.polyfit(V_in_linear, V_out_linear, 1)

print(f"Πειραματική κλίση (Κέρδος Av): {slope:.2f}")
print(f"Πειραματικό offset: {intercept:.2f}")
```
