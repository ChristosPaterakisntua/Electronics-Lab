### Διαγράμματα 5ης Εργαστηριακής Αναφοράς 

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
f = np.array([0.1, 1, 10, 50, 100, 150, 160, 180, 200, 250, 300, 400, 500, 600, ] ) # kHz
V_L = np.array([0.105, 0.5, 0.6, 0.8, 1.29, 1.69, 1.77, 1.93, 2.09, 2.25, 2.41, 2.49, 2.57, 2.57, ]) # V

# συχνότητα αποκοπής
f_c = 159 # kHz

df = pd.DataFrame(
    np.concat([
        f.reshape(-1,1), 
        V_L.reshape(-1, 1)],
    axis=1),
    columns=["f (kHz)", "V_L (V)"],
)
df
```

```python
plt.figure(figsize=(8, 5))

plt.semilogx(df["f (kHz)"], df["V_L (V)"], marker='o', linestyle='-', color='b', label=r"$V_L(f)$")

plt.axvline(f_c, linestyle='--', color="red", alpha=0.5, label=rf"$f_c={f_c}kHz$")

plt.title("Διάγραμμα Τάσης Πηνίου - Συχνότητας f")
plt.xlabel("Συχνότητα f (kHz)")
plt.ylabel("Τάση Πηνίου (V)")
plt.grid(True, which="both", linestyle='--', linewidth=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
f_2 = np.array([0.1, 1, 10, 50, 100, 150, 160, 180, 200, 250, 300, 400, 500, 600, 800, 1000, 1500]) 
V_R = np.array([2.41, 2.41, 2.41, 2.25, 2.09, 1.93, 1.85, 1.77, 1.69, 1.53, 1.37, 1.05, 0.88, 0.72, 0.56, 0.32, 0.15]) # V

df2 = pd.DataFrame(
    np.concat([
        f_2.reshape(-1,1), 
        V_R.reshape(-1, 1)],
    axis=1),
    columns=["f (kHz)", "V_R (V)"],
)

df2
```

```python
plt.figure(figsize=(8, 5))

plt.semilogx(df2["f (kHz)"], df2["V_R (V)"], marker='o', linestyle='-', color='b', label=r"$V_R(f)$")

plt.axvline(f_c, linestyle='--', color="red", alpha=0.5, label=rf"$f_c={f_c}kHz$")

plt.title("Διάγραμμα Τάσης Αντίστασης - Συχνότητας f")
plt.xlabel("Συχνότητα f (kHz)")
plt.ylabel("Τάση Αντίστασης (V)")
plt.grid(True, which="both", linestyle='--', linewidth=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```
