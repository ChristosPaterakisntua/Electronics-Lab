### Διαγράμματα 4ης Εργαστηριακής Αναφοράς 

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
f = np.array([0.01, 0.1, 0.3, 1, 3, 10, 100, ]) # kHz
A_in = np.array([2.69, 2.2, 2.07, 2.05, 2.05, 2.05, 1.99, ]) # V
A_out = np.array([2.73, 2.21, 2.05, 2, 1.73, 0.92, 0.14, ]) # V
```

```python
df = pd.DataFrame(
    np.concat([
        f.reshape(-1,1), 
        A_in.reshape(-1,1), 
        A_out.reshape(-1, 1)],
    axis=1),
    columns=["f (kHz)", "A_in (V)", "A_out (V)"],
)
df["G"] = round(df["A_out (V)"] / df["A_in (V)"], 2)
df
```

```python
plt.figure(figsize=(8, 5))

plt.semilogx(df["f (kHz)"], df["G"], marker='o', linestyle='-', color='b')

plt.title("Διάγραμμα Κέρδους G - Συχνότητας f")
plt.xlabel("Συχνότητα f (kHz)")
plt.ylabel("Κέρδος (G)")
plt.grid(True, which="both", linestyle='--', linewidth=0.5)
plt.show()
```

```python
phi = np.array([1.22, 1.44, 3.52, 11.47, 31.39, 63.89, 87.05, ]) # deg
phi *= -1 

df2 = pd.DataFrame(
    np.concat([f.reshape(-1,1), phi.reshape(-1, 1)], axis=1),
    columns=["f (kHz)", "phi (deg)"],
)

df2
```

```python
plt.figure(figsize=(8, 5))

plt.semilogx(df2["f (kHz)"], df2["phi (deg)"], marker='o', linestyle='-', color='b')

plt.title("Διάγραμμα Φάσης φ - Συχνότητας f")
plt.xlabel("Συχνότητα f (kHz)")
plt.ylabel(r"Φάση φ $(^\circ)$")
plt.grid(True, which="both", linestyle='--', linewidth=0.5)
plt.show()
```
