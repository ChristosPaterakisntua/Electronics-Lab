### Διαγράμματα 1ης Εργαστηριακής Αναφοράς

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
V = np.linspace(-1, 1, 11)
I = np.array([-0.047, -0.037, -0.028, -0.018, -0.009, 0.001, 0.009, 0.018, 0.028, 0.037, 0.047])

df =  pd.DataFrame(np.concat([V.reshape(-1,1), I.reshape(-1,1)], axis=1), columns=["V (V)", "I (mA)"])
df
```

```python
sns.regplot(x=df["V (V)"], y=df["I (mA)"], scatter_kws={"color": "royalblue"}, line_kws={"color": "red"})
plt.title("Χαρακτηριστική V-I", fontdict={"size": 20})
plt.xlabel("V (V)", fontdict={"size": 14})
plt.ylabel("I (mA)", fontdict={"size": 14})
plt.xticks(V)
plt.yticks(np.linspace(-0.06, 0.06, 13))
plt.grid()
plt.tight_layout()
plt.show()
```

```python
df["R (kΩ)"] = df["V (V)"] / df["I (mA)"]
df
```

```python
print(f"Σφάλμα αντίστασης: {(22 - df["R (kΩ)"].sum() / 10) / 22 * 100:.2f}%")
```
